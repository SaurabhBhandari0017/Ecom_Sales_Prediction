```python
from scipy import stats
import numpy as np
```

```python
z_score = np.abs(stats.zscore(df[['unit_price','total_amt']]))
```

```python
df['outliers'] = (z_score>3).any(axis=1)
```

```python
outliers = df[df['outliers']].drop(columns=['outliers'])
cleaned = df[~df['outliers']].drop(columns=['outliers'])
```
